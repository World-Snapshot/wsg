# Dataset Prepare

Place your "Good Dataset" folder into the `wsg/input` folder. Generally speaking, this "Good Dataset" should contain only video mp4 files.

Then:

`python pipeline00_split_video_to_image.py`

This code can split the videos in your dataset into images. You can choose the FPS for separation and the cropping size.

You should also specify set the name of your video dataset in this python code. The `input` folder can contain multiple video datasets, and you should choose one.

I have provided sample processed versions of the test_dataset in the `output` folder. However, it was processed at a **1 fps** rate, and this processing speed depends on the frame rate at which `pipeline00` separates the video. The subsequent pipeline does not take into account the frame rate. The frames separated（crop_frame） out in the output are the files they process.

# Pipeline 12: MonST3R Dynamic Point Cloud Extraction

MonST3R是一个用于从动态视频中提取时变点云的工具，可以生成每帧的相机位姿和内参。相比VGGT，MonST3R具有以下优势：

1. **动态场景支持**：可以处理包含运动物体的视频
2. **时序一致性**：生成的点云在时间上保持连续性
3. **实时模式**：支持快速前馈推理
4. **内存优化**：提供多种内存优化选项

运行前请确保已下载MonST3R模型权重：
```bash
cd monst3r_inference/data && bash download_ckpt.sh
```
