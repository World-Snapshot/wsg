# Dataset Prepare

Put your processed results in the `wsg/output` folder, or reorganize the open-source dataset and then place it in the `wsg/output` folder.

Then:

`python pipeline00_inject_init_wsg_from_crop_frame.py`

This code will create an empty full-channel float16 format WSG file for each video (a single video with the same name in a folder), where the first 3 channels represent the original image values, and the additional 253 channels have values of -1.

You should also specify set the name of your video dataset in this python code. The `input` folder can contain multiple video datasets, and you should choose one.

*I have provided sample processed versions of the test_dataset in the `output` folder. However, it was processed at a **1 fps** rate, and this processing speed depends on the frame rate at which `pipeline00` separates the video. The subsequent pipeline does not take into account the frame rate. The frames separated（crop_frame） out in the output are the files they process.*

The above paragraph you have already read in the extraction script. So now you know that our new processing is only carried out in the `output` folder. The npz files in the `result` folder will be injected into the `wsg` folder under the same name as the video folder.
